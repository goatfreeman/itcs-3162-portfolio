```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
internet_df =pd.read_csv('data/Global Internet Users 1980 - 2020 - kaggle.csv')
```

```python
internet_df
```

```python
internet_df[internet_df['Entity'] == 'Hong Kong']
hong_kong_df = internet_df[internet_df['Entity'] == 'Hong Kong']
hong_kong_df
```

```python
import matplotlib.pyplot as plt
```

```python
hong_kong_df.plot(x='Year',y='No. of Internet Users',kind='line')
```

```python
import seaborn as sns
sns.set(rc={"figure.figsize":(20,12)})
sns.lineplot(data=internet_df,x='Year',y='No. of Internet Users',hue ='Entity')
```

```python
hong_kong_df['No. of Internet Users (Thousands)'] = hong_kong_df['No. of Internet Users'] / 1000
sns.lineplot(data=hong_kong_df, x='Year', y='No. of Internet Users (Thousands)', hue='Entity')
```

```python
hong_kong_data = pd.melt(hong_kong_df, id_vars=['Year'],
                    value_vars=['Internet Users(%)', 'Cellular Subscription', 'Broadband Subscription'],
                    var_name='Metric', value_name='Value')
sns.lineplot(data=hong_kong_data,x='Year',y='Value',hue ='Metric')
```
